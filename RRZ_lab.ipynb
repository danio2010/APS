### Zapiszmy to językiem matematyki

Niech $T(t)$ oznacza temperaturę kawy po $t$ minutach. **Szybkość zmiany** temperatury to pochodna $T'(t)$.
Obserwację „im większa różnica temperatur, tym szybciej kawa stygnie” zapisujemy wzorem:

<div class="rrz-wzor">
<div class="rrz-wzor-podpis">Prawo stygnięcia Newtona</div>

$$
T'(t) = -k\,\big(T(t) - 20\big)
$$

</div>

* $T(t) - 20$ – o ile kawa jest cieplejsza od pokoju,
* $k > 0$ – stała zależna od kubka (im większa, tym szybciej kawa stygnie),
* minus – bo kawa cieplejsza od pokoju **traci** temperaturę, więc pochodna jest ujemna.

<div class="rrz-box rrz-wniosek">
<div class="rrz-box-tytul">Nowe pojęcie</div>

To jest **równanie różniczkowe**: niewiadomą nie jest liczba, tylko **cała funkcja** $T(t)$, a w równaniu występuje jej pochodna.

</div>

<div class="rrz-box rrz-pytanie">
<div class="rrz-box-tytul">Pytanie do klasy</div>

</div>

$$
T(t) = 20 + 70\,e^{-kt}
$$
spełnia równanie $T'(t) = -k\big(T(t)-20\big)$ oraz warunek początkowy $T(0) = 90$.

*Wskazówka:* $\left(e^{-kt}\right)' = -k\,e^{-kt}$.

</div>

Z fizyki znamy drugą zasadę dynamiki Newtona: $\;a = \dfrac{F}{m}$.

Czym jest przyspieszenie $a$? To **szybkość zmiany prędkości**. A prędkość to **szybkość zmiany położenia**. Jeśli $x(t)$ oznacza położenie ciała w chwili $t$, to:

* prędkość: $v(t) = x'(t)$,
* przyspieszenie: $a(t) = v'(t) = x''(t)$ – **druga pochodna** położenia.

<div class="rrz-wzor">
<div class="rrz-wzor-podpis">Druga zasada dynamiki jako równanie różniczkowe</div>

$$
x''(t) = \frac{F}{m}
$$

</div>

<div class="rrz-box rrz-wniosek">
<div class="rrz-box-tytul">Wniosek</div>

**Jeśli znamy siły, możemy przewidzieć ruch.** Cała mechanika – od rzutu piłką po loty na Marsa – opiera się na tym jednym równaniu.

Równanie różniczkowe mówi nam, **jak szybko i w którą stronę zmienia się wielkość w danej chwili**. Dla kawy: jeśli w chwili $t$ ma ona temperaturę $T$, to nachylenie stycznej do wykresu wynosi $T' = -k\,(T - 20)$ – **nie musimy znać wzoru na $T(t)$**, żeby je obliczyć!

Narysujmy więc w wielu punktach $(t, T)$ małe strzałki o takim nachyleniu. Taki rysunek to **pole kierunków**.

<div class="rrz-box rrz-zgadnij">
<div class="rrz-box-tytul">Zadanie · przewidź trajektorię</div>

1. Przesuwaj suwak (albo klikaj **+1 strzałka**) – na rysunku będą pojawiać się kolejne strzałki pola kierunków. Najnowsza jest pomarańczowa.
2. Gdy strzałek będzie dość dużo, **kliknij na wykresie**, w którym miejscu Twoim zdaniem skończy się (w chwili $t = 60$ min) wykres temperatury kawy wychodzący z punktu START.
3. Naciśnij **Sprawdź!** Potem wylosuj nowy start i spróbuj jeszcze raz – tym razem z mniejszą liczbą strzałek.

</div>

```python
k = np.log(7/6) / 5

# Python liczy: strzałki pola kierunków (w losowej kolejności) ...
siatka_t = np.arange(2.5, 60, 5)           # środki kratek co 5 minut
siatka_T = np.arange(5, 100, 10)           # i co 10 stopni
strzalki = [[t0, T0, -k * (T0 - 20)] for t0 in siatka_t for T0 in siatka_T]
np.random.default_rng(2026).shuffle(strzalki)

# ... i prawdziwe trajektorie dla kilku punktów startowych
t = np.linspace(0, 60, 181)
starty = [90, 75, 100, 50, 5, 35, 0, 65]
trajektorie = [20 + (T0 - 20) * np.exp(-k * t) for T0 in starty]

widget_pole_kierunkow(strzalki, t, trajektorie, xlim=(0, 60), ylim=(0, 100), dx=5, dy=10, rownowaga=20,
                      tytul="Pole kierunków równania stygnięcia  T′ = −k(T − 20)")
```

### Pomysł Eulera

Znając temperaturę teraz, znamy też szybkość jej zmiany (z równania). Zróbmy więc mały krok czasu $h$ **wzdłuż stycznej**:

<div class="rrz-wzor">
<div class="rrz-wzor-podpis">Krok metody Eulera</div>

$$
\text{nowa wartość} = \text{stara wartość} + h \cdot \text{pochodna}
$$

</div>

Dla kawy: $\;T_{\text{nowe}} = T_{\text{stare}} + h \cdot \big(-k\,(T_{\text{stare}} - 20)\big)$. Powtarzając ten krok wiele razy, „wędrujemy” w przyszłość.

### Nauczmy tego komputer

Kod jest gotowy – wystarczy uruchomić komórkę. Funkcja `euler` robi $n$ kroków od chwili $0$ do chwili `t_koniec`. Argument `pochodna` to funkcja, która mówi, ile wynosi pochodna dla danej wartości – czyli prawa strona równania różniczkowego.

```python
def euler(pochodna, x0, t_koniec, n):
    h = t_koniec / n          # długość jednego kroku
    t = [0]
    x = [x0]
    for i in range(n):
        x_nowe = x[-1] + h * pochodna(x[-1])     # krok wzdłuż stycznej
        t.append(t[-1] + h)
        x.append(x_nowe)
    return np.array(t), np.array(x)
```

### Wracamy do meteoroidu

Dla ruchu śledzimy **dwie** wielkości naraz – położenie i prędkość – i robimy dla nich kroki Eulera jednocześnie:
* położenie zmienia się z szybkością $v$: $\quad r_{\text{nowe}} = r + h\cdot v$,
* prędkość zmienia się z szybkością $a$: $\quad v_{\text{nowe}} = v + h\cdot a(r)$.

Porównamy dwa modele tego samego zjawiska:
* **model „jabłkowy”**: przyspieszenie zawsze wynosi $g = 9{,}81\ \mathrm{m/s^2}$,
* **model Newtona**: przyspieszenie $a(r) = -\dfrac{G M_Z}{r^2}$.

<div class="rrz-box rrz-zgadnij">
<div class="rrz-box-tytul">Zgadnij przed uruchomieniem</div>

Który model przewidzi szybsze uderzenie? Czy różnica między modelami zależy od tego, **z jakiej wysokości** meteoroid zaczyna spadać?

</div>

```python
GM = 3.986e14          # G * masa Ziemi [m^3/s^2]
R_Z = 6.371e6          # promień Ziemi [m]

def spadek(przyspieszenie, wysokosc, h=0.5):
    """Metoda Eulera dla położenia i prędkości; wysokość w metrach, krok h w sekundach."""
    t, r, v = 0.0, R_Z + wysokosc, 0.0
    czasy, wysokosci = [t], [wysokosc]
    while r > R_Z:                       # dopóki nie uderzy w Ziemię
        a = przyspieszenie(r)
        r, v = r + h * v, v + h * a      # krok Eulera dla obu wielkości
        t = t + h
        czasy.append(t)
        wysokosci.append(r - R_Z)
    return np.array(czasy), np.array(wysokosci), abs(v)

def model_jablkowy(r):
    return -9.81

def model_newtona(r):
    return -GM / r**2

# Python liczy oba modele dla każdej wysokości z suwaka (to chwilę potrwa)
wysokosci_km = [10, 50, 100, 300, 500, 1000, 2000, 3000, 5000, 7500, 10000, 15000, 20000, 30000, 40000]
jablkowy = [spadek(model_jablkowy, H * 1000) for H in wysokosci_km]
newton = [spadek(model_newtona, H * 1000) for H in wysokosci_km]

widget_meteoroid(wysokosci_km, jablkowy, newton, start=wysokosci_km.index(10000),
                 tytul="Spadek meteoroidu: model „jabłkowy” kontra model Newtona")
```

<div class="rrz-box-tytul">Najważniejsza lekcja modelowania</div>

### Lepszy model: populacja z limitem

Załóżmy, że środowisko może utrzymać co najwyżej $M$ ludzi (tzw. **pojemność środowiska**):

<div class="rrz-wzor">
<div class="rrz-wzor-podpis">Model logistyczny</div>

$$
P'(t) = k\,P(t)\left(1 - \frac{P(t)}{M}\right)
$$

</div>

* Gdy $P$ jest małe w porównaniu z $M$, nawias jest bliski $1$ – populacja rośnie jak w modelu wykładniczym.
* Gdy $P$ zbliża się do $M$, nawias jest bliski $0$ – wzrost hamuje.

### Gdy populacje są dwie: zające i lisy

Zające same z siebie się rozmnażają, ale zjadają je lisy. Lisy bez zajęcy wymierają, a im więcej zajęcy, tym lepiej im się wiedzie. Oznaczmy przez $Z(t)$ liczbę zajęcy (w tysiącach), a przez $L(t)$ liczbę lisów (w setkach):

<div class="rrz-wzor">
<div class="rrz-wzor-podpis">Model drapieżnik–ofiara (Lotki–Volterry)</div>

$$
\begin{aligned}
Z'(t) &= a\,Z - b\,Z\,L \qquad &&\text{zające: rozmnażanie minus upolowane}\\
L'(t) &= d\,Z\,L - c\,L &&\text{lisy: przybywa ich, gdy jest co jeść, i umierają}
\end{aligned}
$$

</div>

To **układ dwóch równań**. W każdym punkcie $(Z, L)$ wiemy, jak zmienia się każda z populacji, więc możemy narysować strzałkę: w prawo – zajęcy przybywa, w górę – lisów przybywa. Rozwiązanie to **krzywa fazowa**, która płynie wzdłuż strzałek.
